# DPDP Colab worker

This notebook connects Colab to the persistent coordinator at `dpdp.hari-pi.com`. It does **not** start ngrok or a public server inside Colab.

Before the first run, add `DPDP_WORKER_TOKEN` in Colab's **Secrets** panel (key icon), enable notebook access, select a GPU runtime, and choose **Runtime → Run all**. Keep the cell running while you want Colab to answer queued requests. Never commit this secret.

```python
import pathlib, subprocess
from google.colab import userdata

worker_token = userdata.get('DPDP_WORKER_TOKEN')
if not worker_token:
    raise RuntimeError('Add DPDP_WORKER_TOKEN in Colab Secrets and enable notebook access.')

repo = pathlib.Path('/content/DPDP-Benchmark')
if not (repo / '.git').is_dir():
    subprocess.run([
        'git', 'clone', 'https://github.com/sargunkaur297-droid/DPDP-Benchmark.git', str(repo),
    ], check=True)
else:
    subprocess.run(['git', '-C', str(repo), 'pull', '--ff-only'], check=True)

print('Starting the outbound Colab worker. Leave this cell running.')
subprocess.run(['bash', 'scripts/start_colab_worker.sh'], cwd=repo, check=True)
```
